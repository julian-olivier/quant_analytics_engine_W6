## imports & reading csv

```python
import sys
import os
from src.data_loader import LocalCSVLoader
import src.math_functions as metrics
import matplotlib.pyplot as plt

sys.path.append(os.path.abspath(os.path.join('..')))

loader = LocalCSVLoader(data_directory="./data")
pltr = loader.load_clean_data('PLTR')
```

```python
plt.figure(figsize=(24, 12))
plt.style.use('fivethirtyeight')
plt.plot(pltr.index, pltr['close'])
plt.xlabel('Date')
plt.ylabel('Price ($)')
plt.title('PLTR Stock Price vs Time')
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
